### Voici le notebook que nous mettons à disposition pour tester l'api

#### L'api prend en entrée 3 valeurs:
- owner_name : propriétaire du répertoire
- repo_name  : nom du répertoire
- k          : nombre de répertoires que l'on recommande

L'api ressort un dictionnaire proposant une liste de k urls selon la représentation vectorielle:
- avec l'utilisation du readme
- sans l'utilisation du readme

Les valeurs que nous proposons pour les trois variables sont à titre indicatif, vous pouvez totalement les modifier en faisant attention de requêter sur un répertoire public.  
Il est nécessaire pour le bon fonctionnement de l'API de créer un fichier <b>.env</b> à la racine et d'y ajouter son token d'accès github.  
Si lorsque l'api a été mise en service avec docker compose, le fichier d'environnement ne possédait pas le token d'identification alors il est nécessaire de relancer le docker de l'api pour rendre opérationnelle l'api.
